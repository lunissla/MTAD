Датасет взятий із сайту: https://www.kaggle.com/datasets/saitejabandaruin/imdb-top-700-movies-2026-edition?select=global_tech_startups_2026.csv

```python
import pandas as pd

file_name = r'global_tech_startups_2026.csv'

df = pd.read_csv(
    file_name,
    sep=",",
    na_values='?'
)

df
```

```python
metadata = []

id_columns = ['Company_ID']

target_columns = ['Acquisition_Status']

categorical_columns = [
    'Domain',
    'Country',
    'City',
    'Funding_Stage',
    '2026_Investor_Tier',
    'AI_Adoption_Level',
    'Acquisition_Status'
]

demographic_columns = {
    'Country': 'Geographic',
    'City': 'Geographic'
}

# Опис колонок
descriptions = {
    'Company_ID': 'Унікальний ідентифікатор компанії',
    'Domain': 'Галузь або сфера діяльності стартапу',
    'Founding_Year': 'Рік заснування компанії',
    'Country': 'Країна розташування компанії',
    'City': 'Місто розташування компанії',
    'Funding_Stage': 'Поточна стадія фінансування стартапу',
    'Total_Funding_USD_Millions': 'Загальний обсяг залученого фінансування, млн USD',
    'Valuation_USD_Millions': 'Оцінка вартості компанії, млн USD',
    'Revenue_ARR_Millions': 'Річний регулярний дохід (ARR), млн USD',
    'Monthly_Burn_Rate_Millions': 'Середні щомісячні витрати компанії, млн USD',
    'Runway_Months': 'Кількість місяців, протягом яких компанія може працювати за поточного рівня витрат',
    '2024_Peak_Headcount': 'Максимальна кількість працівників у 2024 році',
    '2023_Layoffs': 'Кількість звільнених працівників у 2023 році',
    '2024_Layoffs': 'Кількість звільнених працівників у 2024 році',
    '2025_Current_Headcount': 'Кількість працівників за даними 2025 року',
    '2026_Investor_Tier': 'Категорія або рівень інвесторів компанії',
    'AI_Adoption_Level': 'Рівень впровадження технологій штучного інтелекту',
    'Acquisition_Status': 'Поточний статус компанії'
}

for column in df.columns:

    if column in id_columns:
        role = 'ID'
    elif column in target_columns:
        role = 'Target'
    else:
        role = 'Feature'

    if column in categorical_columns:
        variable_type = 'Categorical'
    elif pd.api.types.is_integer_dtype(df[column]):
        variable_type = 'Integer'
    elif pd.api.types.is_float_dtype(df[column]):
        variable_type = 'Float'
    else:
        variable_type = 'Categorical'

    demographic = demographic_columns.get(column, 'N/A')

    description = descriptions.get(column, 'N/A')

    missing = 'yes' if df[column].isnull().sum() > 0 else 'no'

    missing_count = df[column].isnull().sum()

    metadata.append({
        'Variable Name': column,
        'Role': role,
        'Type': variable_type,
        'Demographic': demographic,
        'Description': description,
        'Missing Values': missing,
        'Missing Count': missing_count
    })

metadata = pd.DataFrame(metadata)

metadata
```
